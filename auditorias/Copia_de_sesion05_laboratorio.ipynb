> **Pista conceptual:** En Pydantic v2, `model_config = ConfigDict(extra="forbid")` convierte el DTO en un contrato cerrado.

````python

%%writefile app.py

from fastapi import FastAPI
from pydantic import BaseModel, ConfigDict

app = FastAPI(title="API de libros con Pydantic v2")

# Almacenamiento temporal
libros = []
siguiente_id = 1


# DTO de entrada
class LibroEntrada(BaseModel):
    model_config = ConfigDict(extra="forbid")

    titulo: str
    anio: int
    precio: float
    disponible: bool


# DTO de salida
class LibroSalida(BaseModel):
    id: int
    titulo: str
    anio: int
    precio: float
    disponible: bool


# Crear un libro
@app.post(
    "/api/libros",
    response_model=LibroSalida,
    status_code=201
)
async def crear_libro(libro: LibroEntrada):
    global siguiente_id

    nuevo_libro = LibroSalida(
        id=siguiente_id,
        titulo=libro.titulo,
        anio=libro.anio,
        precio=libro.precio,
        disponible=libro.disponible
    )

    libros.append(nuevo_libro)
    siguiente_id += 1

    return nuevo_libro


# Listar libros
@app.get("/api/libros", status_code=200)
async def listar_libros():

    return {
        "libros": libros
    }
````
