## 🧩 El reto: escribe el resultado

Crea el o los archivos del contrato. No hay una plantilla de pasos: el esqueleto solo te da el punto de partida. Todo lo que sigue lo decides tú.

### 📄 `app.py`

Archivo principal de la implementación.

````python

%%writefile app.py

from fastapi import FastAPI
from fastapi import status
from pydantic import BaseModel, Field
import html

app = FastAPI(
    title="API de Comentarios Seguros",
    version="1.0.0"
)

# Modelo para recibir comentarios
class ComentarioEntrada(BaseModel):
    texto: str = Field(min_length=1)

# Almacenamiento temporal en memoria
comentarios = []
siguiente_id = 1


# Registrar un comentario
@app.post(
    "/api/comentarios",
    status_code=status.HTTP_201_CREATED
)
def crear_comentario(entrada: ComentarioEntrada):

    global siguiente_id

    # Escapar el HTML recibido
    texto_seguro = html.escape(entrada.texto)

    # Crear el comentario
    comentario = {
        "id": siguiente_id,
        "texto": texto_seguro
    }

    # Guardar el comentario seguro
    comentarios.append(comentario)

    siguiente_id += 1

    return comentario


# Consultar todos los comentarios
@app.get("/api/comentarios")
def listar_comentarios():

    return {
        "comentarios": comentarios
    }
````
