````python
%%writefile app.py

from fastapi import FastAPI, Query
from pydantic import BaseModel, Field, ConfigDict

app = FastAPI(title="Validación Estricta de Libros")

# Modelo con restricciones de negocio
class LibroCreate(BaseModel):
    model_config = ConfigDict(extra="forbid")

    titulo: str = Field(min_length=3, max_length=80)
    anio: int = Field(ge=1450, le=2100)
    precio: float = Field(gt=0)


# Almacenamiento temporal
libros = []
siguiente_id = 1


# Registrar un libro
@app.post("/api/libros", status_code=201)
def crear_libro(libro: LibroCreate):
    global siguiente_id

    nuevo_libro = {
        "id": siguiente_id,
        "titulo": libro.titulo,
        "anio": libro.anio,
        "precio": libro.precio
    }

    libros.append(nuevo_libro)
    siguiente_id += 1

    return nuevo_libro


# Consultar libros con paginación
@app.get("/api/libros")
def listar_libros(
    limite: int = Query(default=10, ge=1, le=100)
):
    return {
        "libros": libros[:limite],
        "limite": limite
    }
````
