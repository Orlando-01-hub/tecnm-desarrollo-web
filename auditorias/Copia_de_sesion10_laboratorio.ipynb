### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
from fastapi import FastAPI, HTTPException, Response
from pydantic import BaseModel, Field

app = FastAPI(
    title="Mock API de Libros",
    version="0.1.0"
    )

# Implementa aquí el contrato declarado en el reto.
# Tú decides los modelos, las rutas, las validaciones y el manejo de errores.

# Modelo documentado en OpenAPI
class Libro(BaseModel):
    id: int
    titulo: str = Field(min_length=1, max_length=200)
    anio: int = Field(ge=1500, le=2026)

# Modelo para entrada de datos
class LibroEntrada(BaseModel):
    titulo: str = Field(min_length=1, max_length=200)
    anio: int = Field(ge=1500, le=2026)

libros = {}
siguiente_id = 1


@app.get("/api/libros")
def listar_libros():
    return {"libros": list(libros.values())}


@app.post(
    "/api/libros",
    response_model=Libro,
    status_code=201
)
def crear_libro(datos: LibroEntrada):
    global siguiente_id

    libro = {
        "id": siguiente_id,
        "titulo": datos.titulo,
        "anio": datos.anio
    }

    libros[siguiente_id] = libro
    siguiente_id += 1

    return libro


@app.get(
    "/api/libros/{libro_id}",
    response_model=Libro
)
def consultar_libro(libro_id: int):
    if libro_id not in libros:
        raise HTTPException(
            status_code=404,
            detail="Libro no encontrado"
        )

    return libros[libro_id]


@app.put(
    "/api/libros/{libro_id}",
    response_model=Libro
)
def actualizar_libro(libro_id: int, datos: LibroEntrada):
    if libro_id not in libros:
        raise HTTPException(
            status_code=404,
            detail="Libro no encontrado"
        )

    libro = {
        "id": libro_id,
        "titulo": datos.titulo,
        "anio": datos.anio
    }

    libros[libro_id] = libro

    return libro


@app.delete(
    "/api/libros/{libro_id}",
    status_code=204
)
def eliminar_libro(libro_id: int):
    if libro_id not in libros:
        raise HTTPException(
            status_code=404,
            detail="Libro no encontrado"
        )

    del libros[libro_id]

    return Response(status_code=204)


````
